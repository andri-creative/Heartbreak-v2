```python
print("\n📊 STEP 2: HITUNG SUB-SCORE 7 DIMENSI\n")

kolom_B = ['Saya merasa sangat sedih ketika mengingat hubungan yang telah berakhir.', 'Saya merasa kehilangan seseorang yang sangat berarti dalam hidup saya.', 'Saya merasakan sakit emosional ketika memikirkan perpisahan tersebut.', 'Saya merasa hidup saya berubah secara negatif setelah hubungan berakhir.', 'Saya masih merasakan kesedihan yang kuat meskipun waktu telah berlalu.', 'Kenangan tentang hubungan tersebut masih memengaruhi perasaan saya.']

kolom_C = ['Saya masih memiliki keinginan kuat untuk bertemu dengan mantan pasangan.', 'Saya sering ingin menghubungi mantan pasangan', 'Saya sering melihat kembali foto, pesan, atau kenangan bersama mantan.', 'Saya berharap hubungan tersebut dapat kembali seperti dulu.', 'Saya merasa sulit melepaskan keterikatan emosional dengan mantan.', 'Saya masih membayangkan kehidupan bersama mantan.']

kolom_D = ['Saya sering memikirkan alasan mengapa hubungan tersebut berakhir.', 'Saya sering bertanya-tanya apa yang bisa saya lakukan agar hubungan tidak berakhir.', 'Saya sulit menghentikan pikiran tentang mantan.', 'Saya sering mengingat kembali kejadian buruk dalam hubungan.', 'Saya sering menyalahkan diri sendiri atas berakhirnya hubungan.', 'Saya membayangkan kemungkinan hubungan kami kembali.']

kolom_E = ['Perpisahan tersebut membuat saya sulit berkonsentrasi.', 'Saya mengalami perubahan pola tidur setelah putus.', 'Saya kehilangan motivasi melakukan aktivitas sehari-hari.', 'Performa pekerjaan atau studi saya menurun setelah perpisahan.', 'Saya mengurangi interaksi sosial karena perasaan akibat putus cinta.', 'Perpisahan tersebut memengaruhi kesehatan fisik saya.']

kolom_F = ['Saya mulai menerima bahwa hubungan tersebut telah berakhir.', 'Saya dapat menjalani kehidupan tanpa bergantung pada mantan.', 'Saya mampu mengingat hubungan tersebut tanpa rasa sakit yang besar.', 'Saya mulai melihat masa depan dengan lebih positif.', 'Saya merasa siap melanjutkan kehidupan baru.', 'Saya sudah mampu menerima keputusan berakhirnya hubungan.']

kolom_G = ['Saya masih berharap mantan pasangan kembali kepada saya.', 'Saya percaya hubungan kami masih bisa diperbaiki.', 'Saya sulit menerima kemungkinan bahwa hubungan ini benar-benar berakhir.', 'Saya menunggu tanda bahwa mantan masih memiliki perasaan.', 'Saya sulit membuka hati untuk orang lain karena berharap kembali.']

kolom_H = ['Saya memiliki cara sehat untuk mengatasi kesedihan setelah putus.', 'Saya mendapatkan dukungan dari orang lain ketika merasa sedih.', 'Saya mampu mengendalikan emosi ketika mengingat mantan.', 'Saya melakukan aktivitas positif untuk membantu proses pemulihan.', 'Saya mampu menghadapi perasaan kehilangan tanpa merasa kewalahan.']

df['score_B'] = df[kolom_B].mean(axis=1)
df['score_C'] = df[kolom_C].mean(axis=1)
df['score_D'] = df[kolom_D].mean(axis=1)
df['score_E'] = df[kolom_E].mean(axis=1)
df['score_F'] = df[kolom_F].mean(axis=1)
df['score_G'] = df[kolom_G].mean(axis=1)
df['score_H'] = df[kolom_H].mean(axis=1)

print("✅ Sub-score dihitung!")
print(df[['score_B', 'score_C', 'score_D', 'score_E', 'score_F', 'score_G', 'score_H']].describe().round(3))

# ============================================================================
# CELL 2: REVERSE SCORING F & H
print("\n🔄 REVERSE SCORING F & H\n")

df['score_F_reversed'] = 6 - df['score_F']
df['score_H_reversed'] = 6 - df['score_H']

print("✅ Reverse scoring selesai!")
print(f"F: {df['score_F'].mean():.2f} → {df['score_F_reversed'].mean():.2f}")
print(f"H: {df['score_H'].mean():.2f} → {df['score_H_reversed'].mean():.2f}")

# ============================================================================
# CELL 3: HITUNG CRONBACH'S ALPHA (FIXED!)
print("\n🔬 HITUNG CRONBACH'S ALPHA\n")

dimensions = {
    'B': kolom_B,
    'C': kolom_C,
    'D': kolom_D,
    'E': kolom_E,
    'F': kolom_F,
    'G': kolom_G,
    'H': kolom_H
}

for dim, cols in dimensions.items():
    alpha_val, ci = pg.cronbach_alpha(data=df[cols])
    status = "✅" if alpha_val >= 0.70 else "🟡" if alpha_val >= 0.60 else "❌"
    print(f"{dim}: α = {alpha_val:.4f} {status}")

# ============================================================================
# CELL 4: HITUNG HSS
print("\n💔 HITUNG HSS\n")

df['HSS'] = (df['score_B'] + df['score_C'] + df['score_D'] + df['score_E'] + df['score_G']
             - df['score_F_reversed'] - df['score_H_reversed']) / 5

print("✅ HSS dihitung!")
print(df['HSS'].describe().round(3))

# ============================================================================
# CELL 5: BUAT KATEGORI
print("\n📂 BUAT KATEGORI HSS\n")

def categorize_hss(score):
    if score <= 2.33:
        return 'Ringan'
    elif score <= 3.66:
        return 'Sedang'
    else:
        return 'Berat'

df['Kategori_HSS'] = df['HSS'].apply(categorize_hss)

dist = df['Kategori_HSS'].value_counts()
pct = df['Kategori_HSS'].value_counts(normalize=True) * 100

print("✅ Kategori dibuat!")
for kat in ['Ringan', 'Sedang', 'Berat']:
    print(f"{kat}: {dist.get(kat, 0):,} ({pct.get(kat, 0):.1f}%)")

# ============================================================================
# CELL 6: CORRELATION MATRIX
print("\n🔗 CORRELATION MATRIX\n")

corr = df[['score_B', 'score_C', 'score_D', 'score_E', 'score_F_reversed', 'score_G', 'score_H_reversed']].corr()

print("✅ Correlation dihitung!")
print(corr.round(3))

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()

# ============================================================================
# CELL 7: EXPORT HASIL
print("\n💾 EXPORT HASIL\n")

output_path = '/content/drive/MyDrive/STEP2_Hasil_Validasi.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    df[['HSS', 'Kategori_HSS', 'score_B', 'score_C', 'score_D', 'score_E', 'score_F_reversed', 'score_G', 'score_H_reversed']].to_excel(writer, sheet_name='Data Step 2', index=False)
    corr.to_excel(writer, sheet_name='Correlation')

print(f"✅ File disimpan!")
print("\n✅ STEP 2 SELESAI! Siap ke STEP 3")
```

# IMPORT LIBRARY

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from collections import Counter
import pandas as pd
import numpy as np
```

```python
# Target Ke y HSS
y = df['Kategori_HSS']

# Filter x Demografis dengan 8 kolom
x = df[['Umur', 'Jenis Kelamin', 'Pendidikan', 'Lama Hubungan Sebelum Putus',
         'Sudah Berapa Lama Sejak Putus?', 'Siapa yang Mengakhiri Hubungan?',
         'Apakah Masih Berkomunikasi dengan Mantan?', 'Seberapa Sering Melihat Media Sosial Mantan?']].copy()

print("Filter X & Target Y")
print(f"X: {x.shape}")
print(f"Y: {y.shape}")
```

# ONE-HOT ENCODING FITUR KATEGORIK

```python
x_encoded = pd.get_dummies(x, drop_first=True)

print(f"Jumplah Encoded: {x_encoded.shape[1]}")
print(f"Kolom filter: {list(x_encoded.columns)}")
```

# GABUNGKAN DENGAN SUB-SCORE PSIKOMETRIK

```python
x_final = pd.concat([x_encoded, df[['score_B', 'score_C', 'score_D', 'score_E', 'score_F_reversed', 'score_G', 'score_H_reversed']]], axis=1)

print(f"Total fitur (demografis + psikometrik): {x_final.shape[1]}")
print(f"X shape: {x_final.shape}")
print(f"y shape: {y.shape}")
```

# SPLIT TRAIN-TEST

```python
x_train, x_test, y_train, y_test = train_test_split(
    x_final, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(f"X_train: {x_train.shape}")
print(f"X_test: {x_test.shape}")
print(f"y_train: {y_train.shape}")
print(f"y_test: {y_test.shape}")
```

# CLASS IMBALANCE

```python
print("Distribusi di train set:")
train_dist = y_train.value_counts()
train_pct = y_train.value_counts(normalize=True) * 100
for kat in ['Ringan', 'Sedang', 'Berat']:
    count = train_dist.get(kat, 0)
    pct = train_pct.get(kat, 0)
    print(f"  {kat}: {count:,} ({pct:.1f}%)")


print("\nDistribusi di test set:")
test_dist = y_test.value_counts()
test_pct = y_test.value_counts(normalize=True) * 100
for kat in ['Ringan', 'Sedang', 'Berat']:
    count = test_dist.get(kat, 0)
    pct = test_pct.get(kat, 0)
    print(f"  {kat}: {count:,} ({pct:.1f}%)")

# Cek Imbalance
all_balanced = all(15 <= pct <= 50 for pct in train_pct)
if all_balanced:
    print("\n✅ Distribusi seimbang - tidak perlu SMOTE")
    imbalance_flag = False
else:
    print("\n⚠️ Ada imbalance - akan gunakan class_weight di modeling")
    imbalance_flag = True
```

# STANDARDISASI FITUR

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(x_train)
X_test_scaled = scaler.transform(x_test)

# Convert back to DataFrame untuk tracking
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=x_train.columns)
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=x_test.columns)

print("✅ Standardisasi selesai!")
print(f"X_train_scaled mean: {X_train_scaled_df.mean().mean():.6f} (should be ~0)")
print(f"X_train_scaled std: {X_train_scaled_df.std().mean():.6f} (should be ~1)")
```

# Cek Hasil

```python
print(f"\nVariabel yang tersedia:")
print(f"  - X_train_scaled_df: {X_train_scaled_df.shape}")
print(f"  - X_test_scaled_df: {X_test_scaled_df.shape}")
print(f"  - y_train: {y_train.shape}")
print(f"  - y_test: {y_test.shape}")
print(f"  - scaler: StandardScaler object (untuk production)")
print(f"  - imbalance_flag: {imbalance_flag}")
print(f"  - X_train.columns: {len(x_train.columns)} features")
```

# BASELINE MODEL XGBOOST

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
```

# HITUNG CLASS WEIGHTS

```python
from sklearn.utils.class_weight import compute_class_weight

# Hitung class weight untuk handle imbalance
classes = np.unique(y_train)
class_weights = compute_class_weight('balanced', classes=classes, y=y_train)
class_weight_dict = {i: w for i, w in zip(classes, class_weights)}

print("✅ Class weights dihitung:")
for kat, weight in zip(['Ringan', 'Sedang', 'Berat'], class_weights):
    print(f"  {kat}: {weight:.2f}")
```

# TRAIN BASELINE MODEL

```python
# XGBoost baseline
# Label encoding untuk y_train & y_test

X_train_scaled_df.columns = X_train_scaled_df.columns.str.replace('[', '').str.replace(']', '').str.replace('<', '').str.replace('>', '')
X_test_scaled_df.columns = X_test_scaled_df.columns.str.replace('[', '').str.replace(']', '').str.replace('<', '').str.replace('>', '')


label_encoder = {'Ringan': 0, 'Sedang': 1, 'Berat': 2}
y_train_encoded = y_train.map(label_encoder)
y_test_encoded = y_test.map(label_encoder)

xgb_baseline = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    use_label_encoder=False,
    eval_metric='mlogloss',
    n_jobs=-1
)

xgb_baseline.fit(X_train_scaled_df, y_train_encoded)
print(f"Model: XGBClassifier")
print(f"Parameters: n_estimators=100, max_depth=6, learning_rate=0.1")
```

# PREDIKSI DAN EVALUASI

```python
y_pred = xgb_baseline.predict(X_test_scaled_df)

acc = accuracy_score(y_test_encoded, y_pred)
precision = precision_score(y_test_encoded, y_pred, average='weighted', zero_division=0)
recall = recall_score(y_test_encoded, y_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test_encoded, y_pred, average='weighted', zero_division=0)

print(f"Accuracy:  {acc:.4f} ({acc*100:.2f}%)")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
```

# CONFUSION MATRIX

```python
cm = confusion_matrix(y_test_encoded, y_pred)
print("Confusion Matrix:")
print(cm)
print("\nInterpretasi:")
# Use classes variable which correctly reflects the unique classes in y_train
actual_classes = np.unique(y_test.values) # Get the actual class names from y_test

# Map encoded labels back to original class names for interpretation
label_decoder = {0: 'Ringan', 1: 'Sedang', 2: 'Berat'}

for i, kat_encoded in enumerate(actual_classes):
    # Find the original category name using the label_decoder
    original_kat_name = label_decoder[label_encoder[kat_encoded]]

    tp = cm[i, i]
    total = cm[i].sum()
    print(f"  {original_kat_name}: {tp} benar dari {total} ({tp/total*100:.1f}%)")
```

# CLASSIFICATION REPORT

```python
unique_encoded_labels = np.unique(y_test_encoded)
# Map encoded labels back to original class names for target_names
report_target_names = [label_decoder[label] for label in unique_encoded_labels]

print(classification_report(y_test_encoded, y_pred,
                           labels=unique_encoded_labels,
                           target_names=report_target_names))
```

# HYPERPARAMETER TUNING

```python
from sklearn.model_selection import RandomizedSearchCV
import time
```

# DEFINISIKAN PARAMETER GRID

```python
param_dist = {
    'n_estimators': [50, 100, 150, 200, 300],
    'max_depth': [3, 4, 5, 6, 7, 8, 9, 10],
    'learning_rate': [0.01, 0.05, 0.1, 0.15, 0.2],
    'subsample': [0.6, 0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 0.9, 1.0],
    'min_child_weight': [1, 3, 5, 7],
    'gamma': [0, 0.1, 0.5, 1],
}

print(f"Total kombinasi mungkin: {np.prod([len(v) for v in param_dist.values()]):,}")
```

# RANDOMIZEDSEARCHCV

```python
start_time = time.time()

xgb_random = RandomizedSearchCV(
    estimator=XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42),
    param_distributions=param_dist,
    n_iter=50,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42,
    verbose=1
)

xgb_random.fit(X_train_scaled_df, y_train_encoded)

elapsed = time.time() - start_time
print(f"\n✅ Tuning selesai dalam {elapsed/60:.1f} menit")
```

# BEST PARAMETERS

```python
best_params = xgb_random.best_params_
best_score = xgb_random.best_score_

print(f"Best Cross-Validation Accuracy: {best_score:.4f} ({best_score*100:.2f}%)")
print("\nBest Parameters:")
for key, value in best_params.items():
    print(f"  {key:20s}: {value}")
```

# TRAIN FINAL MODEL DENGAN BEST PARAMETERS

```python
xgb_tuned = XGBClassifier(
    **best_params,
    use_label_encoder=False,
    eval_metric='mlogloss',
    random_state=42,
    n_jobs=-1
)

xgb_tuned.fit(X_train_scaled_df, y_train_encoded)
```

# CONFUSION MATRIX DETAILED

```python
cm_final = confusion_matrix(y_test_encoded, y_pred_test)
print("Confusion Matrix:")
print(cm_final)

print("\nPer-Class Performance:")

# Dynamically get the unique encoded labels from y_test_encoded
unique_encoded_labels_test = np.unique(y_test_encoded)

# Use label_decoder to get the original class names for these labels
final_report_class_names = [label_decoder[label] for label in unique_encoded_labels_test]

for i, kat_name in enumerate(final_report_class_names):
    tp = cm_final[i, i]
    fn = cm_final[i].sum() - tp
    fp = cm_final[:, i].sum() - tp

    # Calculate sensitivity (recall)
    sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0

    # Calculate specificity
    tn = np.sum(cm_final) - tp - fn - fp
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

    # Calculate precision
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0

    print(f"\n{kat_name}:")
    print(f"  Sensitivity: {sensitivity*100:.1f}%")
    print(f"  Specificity: {specificity*100:.1f}%")
    print(f"  Precision:   {precision*100:.1f}%")
```

```python
import joblib
import pandas as pd
import numpy as np

# CELL 0: LOAD MODEL
print("᎖ LOAD MODEL\n")

bundle_path = '/content/drive/MyDrive/ANN/Artificial Neural Network/heartbreak_classifier_bundle.pkl'
bundle = joblib.load(bundle_path)

model = bundle['model']
scaler = bundle['scaler']
decode_label = bundle['decode_label']
feature_names = bundle['feature_names']

print("✅ Model loaded!")
print(f"Total features expected: {len(feature_names)}")

# ============================================================================
# CELL 1: FUNCTION PREDICTION (ULTIMATE SYNC)
print("᎑ FUNCTION PREDICTION\n")

def predict_heartbreak(umur, jenis_kelamin, pendidikan, lama_hubungan,
                       sudah_berapa_lama_putus, siapa_yang_mengakhiri,
                       masih_komunikasi, frekuensi_media_sosial_mantan,
                       score_B, score_C, score_D, score_E, score_F, score_G, score_H):

    # 1. Bersihkan label string sebelum dummy encoding
    def sync_clean(val):
        if isinstance(val, str):
            return val.replace('[', '').replace(']', '').replace('<', '').replace('>', '').strip()
        return val

    # 2. Buat DataFrame awal
    demographic_data = pd.DataFrame({
        'Umur': [umur],
        'Jenis Kelamin': [sync_clean(jenis_kelamin)],
        'Pendidikan': [sync_clean(pendidikan)],
        'Lama Hubungan Sebelum Putus': [sync_clean(lama_hubungan)],
        'Sudah Berapa Lama Sejak Putus?': [sync_clean(sudah_berapa_lama_putus)],
        'Siapa yang Mengakhiri Hubungan?': [sync_clean(siapa_yang_mengakhiri)],
        'Apakah Masih Berkomunikasi dengan Mantan?': [sync_clean(masih_komunikasi)],
        'Seberapa Sering Melihat Media Sosial Mantan?': [sync_clean(frekuensi_media_sosial_mantan)]
    })

    # 3. One-hot encode
    X_encoded = pd.get_dummies(demographic_data, drop_first=True)

    # 4. CRITICAL: Bersihkan nama kolom dummy agar tidak ada spasi gantung
    X_encoded.columns = (X_encoded.columns
                         .str.replace('[', '', regex=False)
                         .str.replace(']', '', regex=False)
                         .str.replace('<', '', regex=False)
                         .str.replace('>', '', regex=False)
                         .str.replace('_ ', '_', regex=False) # Hapus spasi setelah underscore
                         .str.strip())

    # 5. Gabungkan dengan psikometri
    psych_scores = pd.DataFrame({
        'score_B': [score_B],
        'score_C': [score_C],
        'score_D': [score_D],
        'score_E': [score_E],
        'score_F_reversed': [6 - score_F],
        'score_G': [score_G],
        'score_H_reversed': [6 - score_H]
    })

    X_input = pd.concat([X_encoded, psych_scores], axis=1)

    # 6. Reindex ke 39 fitur
    X_input = X_input.reindex(columns=feature_names, fill_value=0)

    # 7. Scale & Predict
    X_scaled = scaler.transform(X_input)
    prediction = model.predict(X_scaled)[0]
    prediction_proba = model.predict_proba(X_scaled)[0]

    return {
        'severity': decode_label[prediction],
        'probability': {
            'Ringan': float(prediction_proba[0]),
            'Sedang': float(prediction_proba[1]) if len(prediction_proba) > 1 else 0.0,
            'Berat': float(prediction_proba[2]) if len(prediction_proba) > 2 else 0.0
        },
        'confidence': float(prediction_proba.max())
    }

print("✅ Function ready!")

# ============================================================================
# CELL 2: TEST PREDICTION
print("\n᎕ EXECUTING PREDICTION...\n")

try:
    result = predict_heartbreak(
        umur=25,
        jenis_kelamin='Perempuan',
        pendidikan='S1',
        lama_hubungan='> 5 tahun',
        sudah_berapa_lama_putus='> 1 tahun',
        siapa_yang_mengakhiri='Pasangan yang mengakhiri',
        masih_komunikasi='Kadang-kadang',
        frekuensi_media_sosial_mantan='Kadang-kadang',
        score_B=3.8, score_C=2.9, score_D=4.0, score_E=2.7,
        score_F=3.5, score_G=3.8, score_H=4.0
    )

    print(f"Hasil Prediksi: {result['severity']}")
    print(f"Confidence: {result['confidence']*100:.2f}%")

    print("\nᎉ PROBABILITY:")
    for kat, prob in result['probability'].items():
        bar = '█' * int(prob * 40) + '▒' * (40 - int(prob * 40))
        print(f"  {kat:8s} [{bar}] {prob*100:6.2f}%")
except Exception as e:
    print(f"❌ Error detail: {e}")
```
